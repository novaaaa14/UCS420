# Question 4

```python
import numpy as np

mitali = np.linspace(10, 100, 25)

print("dimensions:", mitali.ndim)

print("shape:", mitali.shape)

print("total elements:", mitali.size)

print("data type:", mitali.dtype)

print("total bytes:", mitali.nbytes)

transpose = mitali.reshape(25, 1)
print("transpose:\n",transpose)

print("transpose (with T):\n",mitali.T)
```

# Question 5

```python
import numpy as np

ucs420_mitali = np.array([
    [10, 20, 30, 40],
    [50, 60, 70, 80],
    [90, 15, 20, 35]
])

print("mean:", np.mean(ucs420_mitali))

print("median:", np.median(ucs420_mitali))

print("maximum:", np.max(ucs420_mitali))

print("minimum:", np.min(ucs420_mitali))

print("unique elements:", np.unique(ucs420_mitali))

reshaped_ucs420_mitali = ucs420_mitali.reshape(4, 3)

print("reshaped:\n",reshaped_ucs420_mitali)

resized_ucs420_mitali = np.resize(ucs420_mitali, (2, 3))

print("resized:\n",resized_ucs420_mitali)
```
