```python
temp = np.array([25, 28, 31, 35, 38, 27, 33, 40])

corrected_temp = temp + 2
print(corrected_temp)

fahrenheit = (corrected_temp * 9/5) + 32

print("temp in fahrenheit:", fahrenheit)

greater_than_32 = corrected_temp[corrected_temp > 32]

print("readings greater than 32C:", greater_than_32)

count = np.sum(corrected_temp > 32)

print("number of readings exceeding 32C:", count)

#Vectorization and Boolean indexing process entire data arrays at once using optimized low-level machine code instead of running slow, item by item loops.
```

# Question 2

```python
steps = np.array([
    [5000, 6200, 7100],
    [8000, 7500, 9000],
    [4500, 5100, 4800],
    [9000, 8500, 9500]
])

print("total steps:", np.sum(steps))

print("mean steps:", np.mean(steps))

print("maximum steps:", np.max(steps))
print("minimum steps:", np.min(steps))

print("total steps for each day:", np.sum(steps, axis=0))

print("total steps for each user:", np.sum(steps, axis=1))

position = np.unravel_index(np.argmax(steps), steps.shape)
print("position of maximum steps:", position)
```

```python
original = np.array([1, 2, 3, 4, 5, 6])

print("original array:", original)

subset = original[1:5]

print("subset:", subset)

subset[0] = 999

print("after modifying subset:")
print("original:", original)
print("subset:", subset)

copied = original[1:5].copy()

copied[0] = 500

print("after modifying copied array:\n")
print("original:", original)
print("copied:", copied)

matrix = np.arange(1, 13).reshape(3, 4)

print("3*4 Matrix:\n")
print(matrix)

print("first row:", matrix[0])

print("last row:", matrix[-1])

print("second column:", matrix[:, 1])

print("rows 1-2 and Columns 2-3:")
print(matrix[0:2, 2:4])

flattened = matrix.flatten()
raveled = matrix.ravel()

print("flattened array:\n", flattened)
print("raveled array:", raveled)

raveled[0] = 100

print("After modifying ravel():\n")
print("raveled array:", raveled)
print("original matrix:")
print(matrix)

flattened[1] = 200

print("after modifying flatten():\n")
print("flattened array:", flattened)
print("original matrix:")
print(matrix)

print("properties of original 3*4 matrix:\n")
print("shape:", matrix.shape)
print("dimensions:", matrix.ndim)
print("size:", matrix.size)
print("data type:", matrix.dtype)
```

# Question 4

```python
y = np.array([40, 65, 30, 85])

X = np.array([
    [6, 70, 3],
    [5, 50, 6],
    [8, 80, 2],
    [4, 30, 8]
])

print("shape of X:", X.shape)
print("dimensions of X:", X.ndim)

print("transpose of X:\n")
print(X.T)

XtX = X.T @ X
print("\nX.T @ X:")
print(XtX)

inverse = np.linalg.inv(XtX)
print("\ninverse of X.T @ X:")
print(inverse)

beta = inverse @ X.T @ y
print("\ncoefficients (Beta):")
print(beta)

new_user = np.array([5, 40, 7])
predicted_score = new_user @ beta

print("\npredicted assistance score:", predicted_score)
```
