```python
roll = 1024160095
```

```python
fixed_entries = [ {"question": "what is the annual fee", "answer": "The annual fee is Rs 500.", "keywords": "fee cost price charge", "category": "billing"},
 {"question": "how to reset password", "answer": "Go to Settings > Reset Password.", "keywords": "password reset login", "category": "account"},
  {"question": "what are your working hours", "answer": "We are open 9 AM to 5 PM.", "keywords": "hours timing open time", "category": "general"},
   {"question": "how can i pay the fee", "answer": "You can pay via UPI, card, or net banking.", "keywords": "pay payment upi fee", "category": "billing"}, ]
```

```python
last_2 = [int(d) for d in str(roll)[-2:] ]
# a = [int(str(roll)[-2:])] can be done like this as well

categories = ["billing","account","general"]

personalized = []
```

```python
for i in last_2:
  category = categories[i%3]

  if category == "billing":
    entry = {
        "question": "how can i check my payment status",
        "answer": "You can check your payment status in the Payments section.",
        "keywords": "payment status transaction billing",
        "category": "billing"
        }
  elif category == "account":
      entry = {
          "question": "how do i update my registered mobile number",
          "answer": "Go to Account Settings and update your mobile number.",
          "keywords": "mobile update account number",
          "category": "account"
        }
  else:
    entry = {
        "question": "how can i contact customer support",
        "answer": "You can contact customer support through email or phone.",
        "keywords": "support contact help service",
        "category": "general"
        }

  personalized.append(entry)
```

```python
all = fixed_entries + personalized

faq_df = pd.DataFrame(all)

print(faq_df)
```

# Question 2

```python
def scoring_func(query,df):
  query1 = query.lower().split()

  res = []

  for i, row in df.iterrows():
    keywords = row["keywords"].lower().split()

    matches = sum(word in keywords for word in query1)

    conf = matches / len(query1)

    if matches>0:
      res.append({
          "question:": row["question"],
          "answer": row["answer"],
          "confidence": conf,
          "category": row["category"]
      })

    res_df = pd.DataFrame(res)

    res_df = res_df.sort_values(
        by="confidence",
        ascending = False
    )

    return res_df
```

```python
query = "payment fee"

res = scoring_func(query,faq_df)

print(res)
```

# Question 3

```python
def same_category(category_name,df):

  res = []

  for i, row in df.iterrows():
    if row["category"] == category_name:
      res.append({
          "question": row["question"],
          "answer": row["answer"]
      })

    return res
```

```python
test_category = "billing"

res = same_category(test_category,faq_df)

print(res)
```

```python
def score_hypothesis(query, df):
  query_w = query.lower().split()

  res = []

  for i, row in df.iterrows():
    keywords = row["keywords"].lower().split()

    matches = sum(word in keywords for word in query_w)

    conf = matches / len(query_w)

    if matches > 0:
      res.append({
          "question": row["question"],
          "answer": row["answer"],
          "confidence": conf,
          "category": row["category"]
          })

  res_df = pd.DataFrame(res)

  res_df = res_df.sort_values(
      by="confidence",
      ascending=False
  )

  if res_df.empty:
    print("No matching entries found")
  else:
    highest = res_df["confidence"].max()

    best_matches = res_df[res_df["confidence"] == highest]

    print("Best matching entry/entries:")
    print(best_matches)

  return res_df
```

```python
score_hypothesis("fee",faq_df)
```
