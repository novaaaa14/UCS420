```python
def isPrime(a):
  if(a<2):
    return False
  elif(a==2):
    return True
  else:
    for i in range(2,int(a**0.5) + 1):
      if(a%i==0):
        return False
      else:
        return True
```

```python
#Question 2:

scores = (90,60,40,34,20,10,10,0)

max = 0
min = 100

max_idx = 0
min_idx = 0

for i in scores:
  if(i>max):
    max = i
    max_idx = scores.index(i)

  if(i<min):
    min = i
    min_idx = scores.index(i)

count = 0

for i in scores:
  if(i==min):
    count+=1

print(max,"index:", max_idx)
print(min, "count:", count, "index:", min_idx)
```

```python
#Question 3

r.seed(1024160095)

L = [r.randint(100, 900) for _ in range(100)]

print(L)
```

```python
count_odd = 0

for i in L:
  if(i%2!=0):
    count_odd+=1
    print(i)
print("odd nums:", count_odd)
```

```python
count_even = 0

for i in L:
  if(i%2==0):
    count_even+=1
    print(i)
print("even nums:", count_even)
```

```python
L_prime = []
count_prime = 0

for i in L:
  if(isPrime(i)):
    count_prime+=1
    L_prime.append(i)

print(count_prime)
print(L_prime)
```

```python
most_frequent_number = None
max_occurrences = 0

processed_nums = []

for i in L:
    if i not in processed_nums:
        current_occurrences = 0
        for j in L:
            if i == j:
                current_occurrences += 1

        if current_occurrences > max_occurrences:
            max_occurrences = current_occurrences
            most_frequent_number = i

        processed_nums.append(j)

print(f"The number that occurs most frequently is: {most_frequent_number}")
print(f"It occurs {max_occurrences} times")
```

```python
#Question 4

roll = [1,0,2,4,1,6,0,0,9,5]

A = []
B = []

for i in roll:
  digit = int(i)
  A.append(digit * 7)

for i in roll:
  digit = int(i)
  B.append(digit * 9)

print(A)
print(B)
```

```python
A = set(A)
B = set(B)
union = A.union(B)
print("Union of A and B:", union)
```

```python
intersection = A.intersection(B)
print("Intersection of A and B:", intersection)
```

```python
difference_A_B = A.difference(B)
difference_B_A = B.difference(A)
print("A - B:", difference_A_B)
print("B - A:", difference_B_A)

#difference gives elements present in one set and not the other
#symmetric difference gives elements that are in either set but not in both
```

```python
symmetric_difference = A.symmetric_difference(B)
print("Symmetric difference of A and B:", symmetric_difference)
```

```python
print("A a subset of B?:", A.issubset(B))
print("B a superset of A?:", B.issubset(A))
```
