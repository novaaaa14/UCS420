```python
def isPrime(a):
  if(a<2):
    return False
  elif(a==2):
    return True
  else:
    for i in range(2,int(a**0.5) + 1):
      if(a%i==0):
        return False
      else:
        return True

a = int(input("enter a: "))
if(isPrime(a)):
  print(a, "is prime")
else:
  print(a, "is not prime")
```
